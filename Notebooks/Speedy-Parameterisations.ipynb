## Let us set up a planet with a land/sea mask and realistic orography. 
We start by running at T31. Can you see where it rains the most?

```julia
using SpeedyWeather

# components
spectral_grid = SpectralGrid(trunc=63, nlayers=8)
large_scale_condensation = ImplicitCondensation(spectral_grid)
convection = SimplifiedBettsMiller(spectral_grid)
drag = NoDrag()
vertical_diffusion = NoVerticalDiffusion(spectral_grid)
surface_wind = NoSurfaceWind(spectral_grid)
surface_heat_flux = NoSurfaceHeatFlux(spectral_grid)

# create model, initialize, run
model = PrimitiveWetModel(spectral_grid; large_scale_condensation, convection, drag, vertical_diffusion, surface_wind, surface_heat_flux)
simulation = initialize!(model)
run!(simulation, period=Day(10))
```

```julia
model
```

```julia
using CairoMakie

(; precip_large_scale, precip_convection) = simulation.diagnostic_variables.physics
precip_large_scale_nodrag=precip_large_scale
m2mm = 1000     # convert from [m] to [mm]
fig=heatmap(m2mm*precip_large_scale_nodrag, title="Large-scale precipiation [mm]: Accumulated over 10 days", colormap=:dense)
save("no_drag_precip.png",fig)
```

```julia
using SpeedyWeather

# components
spectral_grid = SpectralGrid(trunc=63, nlayers=8)
large_scale_condensation = ImplicitCondensation(spectral_grid)
convection = SimplifiedBettsMiller(spectral_grid)
boundary_layer_drag = BulkRichardsonDrag(spectral_grid)
#boundary_layer_drag = LinearDrag(spectral_grid)
vertical_diffusion = BulkRichardsonDiffusion(spectral_grid)
surface_wind = SurfaceWind(spectral_grid)
surface_heat_flux = SurfaceHeatFlux(spectral_grid)

# create model, initialize, run
model = PrimitiveWetModel(spectral_grid; large_scale_condensation, convection, boundary_layer_drag, vertical_diffusion, surface_wind, surface_heat_flux)
simulation = initialize!(model)
run!(simulation, period=Day(10))
```

## Stochastic parameterisation
Now we invoke SP. The scheme will act to perturb the physics tendencies. Look up the discussion here: https://speedyweather.github.io/SpeedyWeatherDocumentation/dev/stochastic_physics/

```julia
using SpeedyWeather

# components
spectral_grid = SpectralGrid(trunc=63, nlayers=8)
random_process = SpectralAR1Process(spectral_grid)
large_scale_condensation = ImplicitCondensation(spectral_grid)
convection = SimplifiedBettsMiller(spectral_grid)
boundary_layer_drag = BulkRichardsonDrag(spectral_grid)
vertical_diffusion = BulkRichardsonDiffusion(spectral_grid)
surface_wind = SurfaceWind(spectral_grid)
surface_heat_flux = SurfaceHeatFlux(spectral_grid)
stochastic_physics = StochasticallyPerturbedParameterizationTendencies(spectral_grid)

# create model, initialize, run
model = PrimitiveWetModel(spectral_grid; large_scale_condensation, convection, boundary_layer_drag, vertical_diffusion, surface_wind, surface_heat_flux, random_process, stochastic_physics)
simulation = initialize!(model)
run!(simulation, period=Day(10))
```

```julia
using CairoMakie

(; precip_large_scale, precip_convection) = simulation.diagnostic_variables.physics
precip_large_scale_SP=precip_large_scale
m2mm = 1000     # convert from [m] to [mm]
fig=heatmap(m2mm*precip_large_scale_SP, title="Large-scale precipiation [mm] with SP: Accumulated over 10 days", colorrange = (0,150), colormap=:dense)
save("SP_precip.png",fig)
```

```julia
# let us plot a few difference fields.

precip_large_scale_SP_diff=precip_large_scale_SP-precip_large_scale_drag
m2mm = 1000     # convert from [m] to [mm]
fig=heatmap(m2mm*precip_large_scale_SP_diff, title="Large-scale precipiation [mm] changes when using SP: Accumulated over 10 days", colorrange = (-25,25), colormap=:bam100)
save("SP_precip-diff.png",fig)
```
